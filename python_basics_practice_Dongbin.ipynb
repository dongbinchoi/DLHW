##2.  Number guessing game

```python
# Number guessing game

import random

secret_number = random.randint(1, 100)
max_tries = 5

print("I'm thinking of a number between 1 and 100.")
print(f"You have {max_tries} tries to guess it!")

for attempt in range(1, max_tries + 1):
    guess = int(input(f"\nTry {attempt}/{max_tries} — Enter your guess: "))

    if guess == secret_number:
        print(f"🎉 Correct! The number was {secret_number}.")
        break
    elif guess < secret_number:
        print("Too low! Try a bigger number.")
    else:
        print("Too high! Try a smaller number.")
else:
    print(f"\n😢 Out of tries! The number was {secret_number}.")
```

## 3. Word guessing game

```python
import random

words = ["fantastic", "brilliant", "great", "fantasy", "smart", "creation", "sophomore", "freshman"]
meanings = ["환상적인", "훌륭한, 명석한", "위대한, 훌륭한", "공상, 환상", "똑똑한", "창조, 창작물", "2학년생", "신입생"]
parts_of_speech = ["adjective", "adjective", "adjective", "noun", "adjective", "noun", "noun", "noun"]

index = random.randint(0, len(words) - 1)
answer = words[index]
meaning = meanings[index]
pos = parts_of_speech[index]

hints = [
    f"Hint 1 — Number of letters: {len(answer)}",
    f"Hint 2 — First letter: {answer[0]}",
    f"Hint 3 — Meaning: {meaning}",
    f"Hint 4 — Part of speech: {pos}"
]

print("I'm thinking of an English word. Try to guess it!")

for i in range(len(hints)):
    print(f"\n{hints[i]}")
    guess = input("Your guess: ")

    if guess.strip().lower() == answer.lower():
        print(f"🎉 Correct! The word was '{answer}'.")
        break
else:
    print(f"\n😢 Out of hints! The word was '{answer}'.")
```

```python
#@markdown Word guessing game
import random
import ipywidgets as widgets
from IPython.display import display, clear_output

words = ["fantastic", "brilliant", "great", "fantasy", "smart", "creation", "sophomore", "freshman"]
meanings = ["환상적인", "훌륭한, 명석한", "위대한, 훌륭한", "공상, 환상", "똑똑한", "창조, 창작물", "2학년생", "신입생"]
parts_of_speech = ["adjective", "adjective", "adjective", "noun", "adjective", "noun", "noun", "noun"]

game_state = {"used_indexes": []}

def start_game():
    clear_output()

    if len(game_state["used_indexes"]) == len(words):
        print("🏆 You've gone through all the words! Great job!")
        restart_button = widgets.Button(description="Start Over 🔄")

        def on_restart_clicked(b):
            game_state["used_indexes"] = []
            start_game()

        restart_button.on_click(on_restart_clicked)
        display(restart_button)
        return

    remaining_indexes = [i for i in range(len(words)) if i not in game_state["used_indexes"]]
    index = random.choice(remaining_indexes)
    game_state["used_indexes"].append(index)

    answer = words[index]
    meaning = meanings[index]
    pos = parts_of_speech[index]

    hints = [
        f"Hint 1 — Meaning: {meaning}",
        f"Hint 2 — First letter: {answer[0]}",
        f"Hint 3 — Part of speech: {pos}",
        f"Hint 4 — Second letter: {answer[1]}",
        f"Hint 5 — Third letter: {answer[2]}"
    ]

    game_state["answer"] = answer
    game_state["hints"] = hints
    game_state["hint_index"] = 0

    output = widgets.Output()
    guess_box = widgets.Text(placeholder="Type your guess here")
    submit_button = widgets.Button(description="Submit ✅")

    def show_next_hint():
        with output:
            print(game_state["hints"][game_state["hint_index"]])

    def on_submit_clicked(b):
        guess = guess_box.value.strip().lower()
        guess_box.value = ""

        with output:
            if guess == game_state["answer"].lower():
                print(f"\n🎉 Correct! The word was '{game_state['answer']}'.")
                show_end_buttons()
                return

        game_state["hint_index"] += 1

        if game_state["hint_index"] < len(game_state["hints"]):
            show_next_hint()
        else:
            with output:
                print(f"\n😢 Out of hints! The word was '{game_state['answer']}'.")
            show_end_buttons()

    def show_end_buttons():
        guess_box.disabled = True
        submit_button.disabled = True
        play_again_button = widgets.Button(description="Play Again 🔁")
        quit_button = widgets.Button(description="Quit ⏹")

        def on_play_again_clicked(b):
            start_game()

        def on_quit_clicked(b):
            clear_output()
            print("Thanks for playing! 👋")

        play_again_button.on_click(on_play_again_clicked)
        quit_button.on_click(on_quit_clicked)
        display(widgets.HBox([play_again_button, quit_button]))

    submit_button.on_click(on_submit_clicked)

    print("I'm thinking of an English word. Try to guess it!")
    display(output, widgets.HBox([guess_box, submit_button]))
    show_next_hint()

start_game()
```

```python
#@markdown Word guessing game with audio

# ============================================================
# ✨ English Vocabulary Guessing Game — Audio Hint Edition ✨
# ============================================================

#!pip install gtts -q

import random
from gtts import gTTS
from IPython.display import display, clear_output, HTML, Audio
import ipywidgets as widgets

words = ["fantastic", "brilliant", "great", "fantasy", "smart", "creation", "sophomore", "freshman"]
meanings = ["환상적인", "훌륭한, 명석한", "위대한, 훌륭한", "공상, 환상", "똑똑한", "창조, 창작물", "2학년생", "신입생"]
parts_of_speech = ["adjective", "adjective", "adjective", "noun", "adjective", "noun", "noun", "noun"]

game_state = {"used_indexes": []}

def speak(text, lang="en"):
    tts = gTTS(text, lang=lang)
    tts.save("response.mp3")
    display(Audio("response.mp3", autoplay=True))

def styled_box(text, color="#2E5591", emoji="💬"):
    display(HTML(f"""
    <div style="background-color:{color}1A; border-left: 5px solid {color};
                padding: 12px 18px; border-radius: 8px; margin: 8px 0;
                font-family: 'Segoe UI', sans-serif; font-size: 16px; color:#222;">
        {emoji} {text}
    </div>
    """))

def start_game():
    clear_output()
    display(HTML("<h2 style='font-family:Segoe UI, sans-serif;'>🎯 Guess the English Word</h2>"))

    if len(game_state["used_indexes"]) == len(words):
        styled_box("You've gone through all the words! Great job! 🏆", color="#27AE60", emoji="🏆")
        speak("You've gone through all the words! Great job!")
        restart_button = widgets.Button(description="Start Over 🔄", button_style="success")

        def on_restart_clicked(b):
            game_state["used_indexes"] = []
            start_game()

        restart_button.on_click(on_restart_clicked)
        display(restart_button)
        return

    remaining_indexes = [i for i in range(len(words)) if i not in game_state["used_indexes"]]
    index = random.choice(remaining_indexes)
    game_state["used_indexes"].append(index)

    answer = words[index]
    meaning = meanings[index]
    pos = parts_of_speech[index]

    # Each hint: (spoken text, language code)
    hints = [
        (meaning, "ko"),
        (f"First letter: {answer[0]}", "en"),
        (f"Part of speech: {pos}", "en"),
        (f"Second letter: {answer[1]}", "en"),
        (answer, "en")  # final hint: full pronunciation of the word
    ]

    game_state["answer"] = answer
    game_state["hints"] = hints
    game_state["hint_index"] = 0

    output = widgets.Output()
    guess_box = widgets.Text(placeholder="Type your guess here...", layout=widgets.Layout(width="300px"))
    submit_button = widgets.Button(description="Submit ✅", button_style="primary")

    def show_next_hint():
        hint_number = game_state["hint_index"] + 1
        is_last = hint_number == len(game_state["hints"])
        text, lang = game_state["hints"][game_state["hint_index"]]

        with output:
            if is_last:
                styled_box(f"Hint {hint_number} — 🔊 Listen closely: this is how the word sounds.", color="#2E5591", emoji="🔊")
            else:
                styled_box(f"Hint {hint_number} — 🔊 Listen for a clue.", color="#2E5591", emoji="🔊")
        speak(text, lang=lang)

    def on_submit_clicked(b):
        guess = guess_box.value.strip().lower()
        guess_box.value = ""

        if guess == game_state["answer"].lower():
            with output:
                styled_box(f"Correct! The word was '{game_state['answer']}'.", color="#27AE60", emoji="🎉")
            speak(f"Correct! The word was {game_state['answer']}")
            show_end_buttons()
            return

        game_state["hint_index"] += 1

        if game_state["hint_index"] < len(game_state["hints"]):
            with output:
                styled_box("Not quite — here's another hint.", color="#E67E22", emoji="❌")
            show_next_hint()
        else:
            with output:
                styled_box(f"Out of hints! The word was '{game_state['answer']}'.", color="#C0392B", emoji="😢")
            speak(f"Out of hints! The word was {game_state['answer']}")
            show_end_buttons()

    def show_end_buttons():
        guess_box.disabled = True
        submit_button.disabled = True
        play_again_button = widgets.Button(description="Play Again 🔁", button_style="info")
        quit_button = widgets.Button(description="Quit ⏹", button_style="danger")

        def on_play_again_clicked(b):
            start_game()

        def on_quit_clicked(b):
            clear_output()
            styled_box("Thanks for playing! 👋", color="#2E5591", emoji="👋")
            speak("Thanks for playing!")

        play_again_button.on_click(on_play_again_clicked)
        quit_button.on_click(on_quit_clicked)
        display(widgets.HBox([play_again_button, quit_button]))

    submit_button.on_click(on_submit_clicked)

    display(output, widgets.HBox([guess_box, submit_button]))
    show_next_hint()

start_game()
```

## 4. Date and time

```python
from datetime import datetime
import pytz

city_timezones = {
    "Seoul": "Asia/Seoul",
    "New York": "America/New_York",
    "London": "Europe/London",
    "Tokyo": "Asia/Tokyo"
}

city = input("Enter a city name (Seoul, New York, London, Tokyo): ")

if city in city_timezones:
    tz = pytz.timezone(city_timezones[city])
    current_time = datetime.now(tz)
    print(f"The current date and time in {city} is: {current_time.strftime('%Y-%m-%d %H:%M:%S')}")
else:
    print("Sorry, that city is not in the list.")
```

```python
#@markdown Date and time in audio
# ============================================================
# 🌍 City Date & Time — Voice Response Edition
# ============================================================

#!pip install gtts pytz -q

from datetime import datetime
import pytz
from gtts import gTTS
from IPython.display import Audio, display

city_timezones = {
    "Seoul": "Asia/Seoul",
    "New York": "America/New_York",
    "London": "Europe/London",
    "Tokyo": "Asia/Tokyo"
}

def speak(text):
    tts = gTTS(text)
    tts.save("response.mp3")
    display(Audio("response.mp3", autoplay=True))

city = input("Enter a city name (Seoul, New York, London, Tokyo): ")

if city in city_timezones:
    tz = pytz.timezone(city_timezones[city])
    current_time = datetime.now(tz)
    formatted_time = current_time.strftime("%B %d, %Y at %I:%M %p")

    print(f"The current date and time in {city} is: {formatted_time}")
    speak(f"The current date and time in {city} is {formatted_time}")
else:
    message = "Sorry, that city is not in the list."
    print(message)
    speak(message)
```
